```python
import pandas as pd

# Chemin vers ton fichier brut
chemin = "../01_donnees_brutes/suivi_lots_2021.xlsx"

# On ouvre le classeur sans lire les données, juste pour l'inspecter
classeur = pd.ExcelFile(chemin)

print("Nombre de feuilles :", len(classeur.sheet_names))
print("\nListe des feuilles :")
for nom in classeur.sheet_names:
    print(" -", nom)
```

```python
# On lit UNE feuille pour examiner sa structure interne.
# header=None : on NE suppose PAS que la 1re ligne est un en-tête,
# car ces feuilles ont souvent des titres avant le vrai tableau.
feuille = pd.read_excel(chemin, sheet_name="5608", header=None)

print("Dimensions (lignes, colonnes) :", feuille.shape)

# On affiche les 15 premières lignes pour voir comment c'est organisé
feuille.head(15)
```

```python
# header=3 : on dit à pandas que la ligne d'indice 3 contient les noms de colonnes
feuille = pd.read_excel(chemin, sheet_name="5608", header=3)

print("Colonnes détectées :")
print(list(feuille.columns))

feuille.head(12)
```

```python
# Dictionnaire : ancien nom  ->  nouveau nom (plus simple)
colonnes_utiles = {
    "BLOC": "bloc",
    "LOT": "lot",
    "SUPERFICIE": "superficie",
    "OCCUPEE": "occupee",
    "RESTANTE": "restante",
    "NOMS ET PRENOMS DE CLIENTS": "client",
    "OBSERVATIONS": "observation",
}

# On garde uniquement ces colonnes, puis on les renomme
feuille = feuille[list(colonnes_utiles.keys())].rename(columns=colonnes_utiles)

feuille.head(12)
```

```python
# .ffill() = forward fill : chaque case vide de "bloc" est remplie
# avec la dernière valeur de bloc rencontrée au-dessus.
feuille["bloc"] = feuille["bloc"].ffill()

feuille.head(12)
```

```python
# 1) Combien de valeurs manquantes (NaN) dans chaque colonne ?
print("Valeurs manquantes par colonne :")
print(feuille.isna().sum())

print("\nNombre total de lignes :", len(feuille))

# 2) Les dernières lignes : on y trouve souvent des totaux ou des lignes vides
feuille.tail(15)
```

```python
# On convertit lot et superficie en nombres.
# errors="coerce" : ce qui n'est PAS un nombre (texte "LOTS", "TOTAL BLOC 19"...)
# devient automatiquement NaN (vide).
feuille["lot"] = pd.to_numeric(feuille["lot"], errors="coerce")
feuille["superficie"] = pd.to_numeric(feuille["superficie"], errors="coerce")

# On garde seulement les lignes où lot ET superficie sont des nombres valides
avant = len(feuille)
feuille = feuille.dropna(subset=["lot", "superficie"])

print(f"Avant : {avant} lignes")
print(f"Après : {len(feuille)} lignes")
print(f"Supprimées (totaux, en-têtes, vides) : {avant - len(feuille)}")

feuille.tail(15)
```

```python
# On convertit le bloc en nombre entier.
# Int64 (avec majuscule) est un type entier qui tolère les valeurs vides.
feuille["bloc"] = pd.to_numeric(feuille["bloc"], errors="coerce").astype("Int64")

# Vérification : combien de parcelles par bloc dans cette feuille ?
print("Nombre de parcelles par bloc :")
print(feuille["bloc"].value_counts().sort_index())
```

```python
# On s'assure que "restante" est bien numérique
feuille["restante"] = pd.to_numeric(feuille["restante"], errors="coerce")

# Règle métier :
#   restante = 0   -> parcelle payée      -> "Vendu"
#   restante > 0   -> reste à commercialiser -> "Disponible"
feuille["statut"] = feuille["restante"].apply(
    lambda r: "Vendu" if r == 0 else "Disponible"
)

# Vérification : la répartition vendus / disponibles
print(feuille["statut"].value_counts())

# Aperçu avec les colonnes clés
feuille[["bloc", "lot", "superficie", "restante", "client", "statut"]].head(12)
```

```python
# 1) D'où viennent ces parcelles ? On marque le titre foncier.
feuille["TF"] = "5608"

# 2) Version numérique de la cible : 1 si vendu, 0 sinon
feuille["vendu"] = (feuille["statut"] == "Vendu").astype(int)

# 3) On garde et ordonne les colonnes finales
colonnes_finales = ["TF", "bloc", "lot", "superficie", "restante", "client", "statut", "vendu"]
feuille_5608 = feuille[colonnes_finales].reset_index(drop=True)

print("Forme finale :", feuille_5608.shape)
feuille_5608.head()
```

```python
def nettoyer_feuille(nom_feuille):
    # 1) Lecture brute pour repérer la ligne d'en-tête (celle qui contient BLOC et LOT)
    brut = pd.read_excel(chemin, sheet_name=nom_feuille, header=None)
    ligne_entete = None
    for i in range(min(15, len(brut))):
        ligne = brut.iloc[i].astype(str).str.upper()
        if ligne.str.contains("BLOC").any() and ligne.str.contains("LOT").any():
            ligne_entete = i
            break
    if ligne_entete is None:
        return None   # pas une feuille de parcelles -> on l'ignore

    # 2) Relecture avec le bon en-tête
    df = pd.read_excel(chemin, sheet_name=nom_feuille, header=ligne_entete)

    # 3) Renommer les colonnes utiles en cherchant des mots-clés
    renom = {}
    for col in df.columns:
        c = str(col).upper()
        if "BLOC" in c: renom[col] = "bloc"
        elif "LOT" in c: renom[col] = "lot"
        elif "SUPERFICIE" in c: renom[col] = "superficie"
        elif "RESTANTE" in c: renom[col] = "restante"
        elif "CLIENT" in c or "NOMS" in c: renom[col] = "client"
    df = df.rename(columns=renom)

    # 4) Vérifier qu'on a bien les colonnes essentielles
    if not all(c in df.columns for c in ["bloc", "lot", "superficie", "restante"]):
        return None

    # 5) Nettoyage (exactement nos étapes manuelles)
    df["bloc"] = df["bloc"].ffill()
    for c in ["bloc", "lot", "superficie", "restante"]:
        df[c] = pd.to_numeric(df[c], errors="coerce")
    df = df.dropna(subset=["bloc", "lot", "superficie"])

    # 6) Statut + cible + titre foncier
    df["statut"] = df["restante"].apply(lambda r: "Vendu" if r == 0 else "Disponible")
    df["vendu"] = (df["statut"] == "Vendu").astype(int)
    df["TF"] = nom_feuille
    if "client" not in df.columns:
        df["client"] = None

    # 7) Colonnes finales
    return df[["TF","bloc","lot","superficie","restante","client","statut","vendu"]].reset_index(drop=True)


# On teste la fonction sur la feuille qu'on connaît déjà : 5608
test = nettoyer_feuille("5608")
print("Forme obtenue :", test.shape)
test.head()
```

```python
def nettoyer_feuille(nom_feuille):
    # 0) NOUVEAU : exclure d'emblée les récaps / totaux / situations
    nom_maj = nom_feuille.upper()
    if any(mot in nom_maj for mot in ["RECAP", "DISPONIBIL", "TOTAL", "SITUATION"]):
        return None

    # 1) Trouver la ligne d'en-tête
    brut = pd.read_excel(chemin, sheet_name=nom_feuille, header=None)
    ligne_entete = None
    for i in range(min(15, len(brut))):
        ligne = brut.iloc[i].astype(str).str.upper()
        if ligne.str.contains("BLOC").any() and ligne.str.contains("LOT").any():
            ligne_entete = i
            break
    if ligne_entete is None:
        return None

    # 2) Relire avec le bon en-tête
    df = pd.read_excel(chemin, sheet_name=nom_feuille, header=ligne_entete)

    # 3) Renommer par mots-clés (CORRIGÉ : SUPERF, RESTANT, OCCUPANT)
    renom = {}
    for col in df.columns:
        c = str(col).upper()
        if "BLOC" in c: renom[col] = "bloc"
        elif "LOT" in c: renom[col] = "lot"
        elif "SUPERF" in c: renom[col] = "superficie"
        elif "RESTANT" in c: renom[col] = "restante"          # capte RESTANT et RESTANTE
        elif "CLIENT" in c or "NOMS" in c or "OCCUPANT" in c: renom[col] = "client"
    df = df.rename(columns=renom)

    # 4) Vérifier les colonnes essentielles
    if not all(c in df.columns for c in ["bloc", "lot", "superficie", "restante"]):
        return None

    # 5) Nettoyage
    df["bloc"] = df["bloc"].ffill()
    for c in ["bloc", "lot", "superficie", "restante"]:
        df[c] = pd.to_numeric(df[c], errors="coerce")
    df = df.dropna(subset=["bloc", "lot", "superficie"])

    # 6) Statut + cible + TF
    df["statut"] = df["restante"].apply(lambda r: "Vendu" if r == 0 else "Disponible")
    df["vendu"] = (df["statut"] == "Vendu").astype(int)
    df["TF"] = nom_feuille
    if "client" not in df.columns:
        df["client"] = None

    # 7) Colonnes finales
    return df[["TF","bloc","lot","superficie","restante","client","statut","vendu"]].reset_index(drop=True)


# On relance la boucle sur toutes les feuilles
tables, ignorees = [], []
for nom in classeur.sheet_names:
    try:
        resultat = nettoyer_feuille(nom)
        if resultat is not None and len(resultat) > 0:
            tables.append(resultat)
            print(f"OK    {nom:32} -> {len(resultat):3} parcelles")
        else:
            ignorees.append(nom)
    except Exception as e:
        ignorees.append(nom)

print(f"\nFeuilles exploitées : {len(tables)} | ignorées : {len(ignorees)}")
```

```python
# Combien de parcelles au total jusqu'ici ?
total = sum(len(t) for t in tables)
print("Total parcelles récupérées :", total)

# Quelles feuilles sont encore ignorées ?
print("\nFeuilles encore ignorées :")
for nom in ignorees:
    print(" -", nom)
```

```python
tf_manques = ["3638", "3905 nvo", "5176 nvo", "5066", "5608SM"]

for nom in tf_manques:
    brut = pd.read_excel(chemin, sheet_name=nom, header=None)
    trouve = False
    # On cherche la ligne qui contient "BLOC" dans les 20 premières
    for i in range(min(20, len(brut))):
        ligne = brut.iloc[i].astype(str).str.upper()
        if ligne.str.contains("BLOC").any():
            colonnes = [str(v) for v in brut.iloc[i].tolist() if str(v) != "nan"]
            print(f"{nom:10} | en-tête ligne {i} : {colonnes}")
            trouve = True
            break
    if not trouve:
        print(f"{nom:10} | aucun 'BLOC' dans les 20 premières lignes")
```

```python
# pd.concat empile toutes les tables de la liste, les unes sous les autres
base = pd.concat(tables, ignore_index=True)

print("Base consolidée :", base.shape)
print("Nombre de lotissements (TF) :", base["TF"].nunique())
print("Taux de commercialisation global :", round(base["vendu"].mean()*100, 1), "%")

print("\nRépartition vendu / disponible :")
print(base["statut"].value_counts())

print("\nParcelles par lotissement :")
print(base["TF"].value_counts())

base.head()
```

```python
# 1) Retirer la feuille parasite (nom de personne, 1 seule parcelle)
base = base[base["TF"] != "KOMGUEM KAINNENG NADEGE"].reset_index(drop=True)

# 2) Vérification finale
print("Base finale :", base.shape)
print("Lotissements :", base["TF"].nunique())
print("Vendu / Disponible :", base["vendu"].sum(), "/", (base["vendu"] == 0).sum())

# 3) Export vers le dossier des données propres
base.to_csv("../02_donnees_propres/base_parcelles.csv", index=False)
base.to_excel("../02_donnees_propres/base_parcelles.xlsx", index=False)

print("\nBase exportée dans 02_donnees_propres/ :")
print(" - base_parcelles.csv")
print(" - base_parcelles.xlsx")
```
